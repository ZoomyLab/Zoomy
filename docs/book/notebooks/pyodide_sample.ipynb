```python
import math

g = 9.81
h = 1.0
u = 1.5

c = math.sqrt(g * h)
froude = abs(u) / c

print(f"gravity wave speed c = {c:.4f} m/s")
print(f"Froude number Fr = {froude:.4f}")
```
